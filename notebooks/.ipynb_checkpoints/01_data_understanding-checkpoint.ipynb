# 01. Data Understanding - Berlin Housing Market Analysis
**Objectives**
Explore the structure and basic characteristis of the Berlin housing

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')
plt.rcParams['figure.figsize']= (10, 6)

rentals= pd.read_csv('../data/raw/rentals.csv')

print("Rentals dataset loaded successfully")
print("Shape:", rentals.shape)
print("\nColumns:")
print(rentals.columns.tolist())
```

```python
print("===Basic Info ===")
print(rentals.info())
```

```python
print("=== First 5 rows ===")
rentals.head() 
```

```python
print("=== Missing Values ===")
missing = rentals.isnull().sum()
print(missing[missing >0])
print("\nTotal missing values:")
```

```python
print("=== Numerical Summary ===")
rentals.describe().round(2) 
```

```python
print("=== Unique values in key categorical columns ===")
print("Districts (Bezirk):", rentals['bezirk'].nunique())
print(rentals['bezirk'].value_counts().head(10))

print("\nProperty Type:")
print(rentals['property_type'].value_counts())

print("\nKiez Premium:")
print(rentals['kiez_premium'].value_counts())
```

```python
print("=== Rent per m2 (Kalt) ===")
print(rentals['rent_per_m2_kalt_eur'].describe().round(2))
```

### Initial Observations
- The dataset is clean (no missing values).
- Average kaltmiete is around 947 euros with an average size of 76m2.
- Average rent per m2 is app 12.12 euros.
- Reinikendorf, Treptow and Pankow have the highest number of listings.
- There is a wide range in both sieze and rent due to luxury apartments.
