# 02. Data Cleaning - Berlin Housing Market Analysis

**Objectives**
Clean and prepare the rentals dataset for analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
sns.set_style('whitegrid')

rentals = pd.read_csv('../data/raw/rentals.csv')

print("Data loaded successfully")
print("Shape:", rentals.shape) 
```

```python
print("=== Data Types ===")
print(rentals.dtypes)

print("\n=== Checking for duplicates ===")
print("Number of duplicate rows:")
```

```python
rentals['date_listed'] = pd.to_datetime(rentals['date_listed'])

print("Earliest date:", rentals['date_listed'].min())
print("Latest date:", rentals['date_listed'].max())
```

```python
rentals['year'] = rentals['date_listed'].dt.year
rentals['month']= rentals['date_listed'].dt.month

rentals['year_month'] = rentals['date_listed'].dt.to_period('M')

print("New time columns created:")
print(rentals[['date_listed', 'year', 'month', 'year_month']].head()) 
```

```python
print("=== Extreme Values Check ===")
print("Min rent per m2:", rentals['rent_per_m2_kalt_eur'].min())
print("Max rent per m2:", rentals['rent_per_m2_kalt_eur'].max())
print("Min area:", rentals['area_m2'].min())
print("Max area:", rentals['area_m2'].max()) 
```

```python
useful_cols = [
    'id', 'date_listed', 'year', 'month', 'year_month', 'ortsteil', 'bezirk', 'kiez_premium', 'property_type', 'rooms', 'area_m2', 'floor', 'total_floors', 'building_era', 'energy_class','has_lift', 'has_balcony', 'furnished', 'to_brandenburg_gate_km','kaltmiete_eur_monthly', 'nebenkosten_eur_monthly', 'warmmiete_eur_monthly', 'rent_per_m2_kalt_eur'
              ]

rentals_clean = rentals[useful_cols].copy()
print("Cleaned dataset shape:", rentals_clean.shape) 
```

```python
rentals_clean.to_csv('../data/processed/rentals_cleaned.csv', index=False)
print("Cleaned data saved successfully") 
```
