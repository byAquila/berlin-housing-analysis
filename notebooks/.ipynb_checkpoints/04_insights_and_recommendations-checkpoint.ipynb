#04. Insights and Recommendations - Berlin Housing Market Analysis

** Project Goals**
Analyze Berlin rental market to uncover key patterns and provide actionable insights.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

rentals = pd.read_csv('../data/processed/rentals_cleaned.csv')
rentals['date_listed'] = pd.to_datetime(rentals['date_listed'])

print("=== Project Summary ===")
print("Total listings analyzed:", len(rentals))
print("Date range:", rentals['date_listed'].min().date(), rentals['date_listed'].max().date())
print("Average cold rent: Eur", round(rentals['kaltmiete_eur_monthly'].mean(), 2))
print("Average rent per m2: Eur", round(rentals['rent_per_m2_kalt_eur'].mean(), 2))
```

### Key Insights
1. **Rising Rental Prices**
     Average rent per m2 has increased over time and surpassed 14 eur by late 2025. This shows strong upward pressure on rents in Berlin.

2. **District Differences**
     Friedrichshain-Kreuzberg has the highest average rents. Other expensive areas tend to be central and well-connected districts.

3. **Size vs Price**
     Smaller apartments (especially under 100m2)often have higher rent per square meter. Larger apartments cost more in total but are usually cheaper per m2.

4. **Supply Distribution**
    Reinickendorf, Treptow-Kopenick, and Pankow have the highest number of listings, indicating higher supply in these areas.

5. **Clean Dataset**
    The dataset contained no missing values and very few extreme outliers, making the analysis reliable.


### Recommendations

** For Renters**
- Consider districts with higher supply (eg, Reinickendorf, Treptow-Kopenick, Pankow) for better probality.
- Smaller apartments are more expensive per m2 - calculate total monthly cost carefully.
- Act faster in popular districts like Friedrichshain as demand is high.

** For Analysts / Investors**
- Monitor the upward trend in rent per m2 closely.
- Central and well-connected districts continue to command premium prices.
- Neighborhood-level analysis (Ortsteil) gives more precise insights than district level alone.

**Possible Future Analysis**
- Compare rentals with secondary sales and new construction prices.
- Analyze the impact of energy class, balcony, and lift on rental prices.
- Study seasonal patterns in listings.



### Conclusion

This project analyzed 30,000 rental listings in Berlin from 2020 to 2026.
    The main finding is a clear increase in rental prices over time, with significant differences between districts and apartment sizes.
